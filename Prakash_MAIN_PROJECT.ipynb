**Dataset Column/features Description:**

* **Claim_ID**-	Unique identification number assigned to each healthcare claim.
* **Name**-	Name of the patient associated with the claim.
* **Age**-	Age of the patient at the time of the claim.
* **Gender**-	Gender of the patient.
* **Medical Condition**-	Primary medical condition associated with the patient/claim.
* **CPT_HCPCS_Code**-	Medical procedure or service code used for billing.
* **Procedure_Description**-	Description of the medical procedure or service performed.
* **Provider_ID**-	Unique identification number of the healthcare provider.
* **Provider_Name**-	Name of the healthcare provider associated with the claim.
* **Facility_Name**-	Name of the healthcare facility where the service was provided.
* **Payer_Name**-	Name of the insurance company or payer responsible for the claim.
* **Payer_Type**-	Type of payer, such as Commercial or Government.
* **State**-	U.S. state associated with the claim or healthcare service.
* **Region**-	Geographic region associated with the state.
* **Date_of_Admission**-	Date on which the patient was admitted to the healthcare facility.
* **Discharge_Date**-	Date on which the patient was discharged from the facility.
* **Claim_Submission_Date**-	Date on which the claim was submitted to the payer.
* **Claim_Month**-	Month in which the claim was submitted.
* **Claim_Year**-	Year in which the claim was submitted.
* **Claim_Status**-	Current status of the claim, such as Paid or Denied.
* **Billed_Amount**-	Total amount initially billed by the healthcare provider.
* **Allowed_Amount**-	Amount approved or allowed by the payer according to the applicable contract or fee schedule.
* **Contractual_Adjustment**-	Difference between the billed amount and the allowed amount due to contractual agreements.
* **Paid_Amount**-	Amount actually paid by the payer for the claim.
* **Outstanding_Amount**-	Remaining amount that has not been paid or settled.
* **Denial_Code**-	Standard code identifying the reason for claim denial.
* **Denial_Reason**-	Description explaining why the claim was denied.
* **Denial_Category**-	Broad classification of the claim denial, such as Contractual or other denial categories.
* **Denial_Amount**-	Financial amount associated with the denied portion of the claim.
* **Posting_Rep**-	Payment posting representative responsible for processing/posting the claim payment.
* **Payment_Posting_Date**-	Date on which the payment was posted to the claim.
* **Admission_Type**-	Type of hospital admission, such as Emergency, Urgent, or Elective.
* **DOS (Date of Service)**-	The date on which the healthcare service or procedure was provided to the patient. It is used as the base date for calculating claim aging.
* **Days_Outstanding**-	Number of days the claim or outstanding amount has remained unpaid from the Date of Service (DOS). It helps measure the age of outstanding AR.
* **AR_Aging_Bucket**-	Categorizes outstanding claims based on their age, such as 0–30, 31–60, 61–90, 91–120, and 120+ days.
* **Assigned_AR_Rep**-	Identifies the AR representative responsible for working and following up on the claim.
* **Priority**-	Indicates the priority level of the claim based on factors such as outstanding amount and claim age: High, Medium, or Low.
* **Follow-up_Status**-	Shows the current status of AR follow-up activity: Not Started, In Progress, or Escalated.
* **Appeal_Status**-	Indicates the status of an appeal for a denied claim: Filed, In Progress, Won, or Lost.
* **Last_Follow-up_Date**-	The most recent date on which the AR representative performed a follow-up activity on the claim.
* **Next_Action**-	Specifies the next action required to work the claim, such as Call Payer, Refile, Appeal, or Write-off.
* **Escalation_Flag**-	Identifies claims requiring escalation, typically because they have remained unresolved or outstanding beyond the defined aging threshold.
* **Resolution_Amount**-	The amount recovered or resolved as a result of AR follow-up, payment, appeal, or other resolution activity.
* **Remaining_Outstanding**-	The amount that remains unpaid after the resolution or follow-up activity.

```python
# Identify Duplicates
df.duplicated().sum()
```

```python
# Remove Duplicate Records

df = df.drop_duplicates()

df.duplicated().sum()
```

```python
# Name Correction

df.loc[:, "Name"] = df["Name"].str.title()

df["Name"]
```

```python
# Age – Mean and Median Analysis

Mean = df["Age"].mean()
print(Mean)

Median = df["Age"].median()
print(Median)

df["Age"].isnull().sum()
```

```python
# Age – Replace Null Values using Median

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Age"].isnull().sum()

df["Age"] = df["Age"].astype(int)

df["Age"].dtype
```

```python
# Gender – Replace incorrect Values

df["Gender"] = df["Gender"].replace({

   "Male"   : "M",
   "Female" : "F"
})

print(df.Gender)
#print(df.Gender.nunique())

```

```python
# Medical Condition – Analyze the missing values based on CPT_HCPCS_Code, as this column represents the procedure corresponding to the Medical condition.

df.loc[df["Medical Condition"].isnull(),"CPT_HCPCS_Code"].value_counts()

a = df.groupby("CPT_HCPCS_Code")["Medical Condition"]
print(a.count())
```

```python
# Medical Condition – Replace Null Values Using the Mode Based on CPT_HCPCS_Code.


df["Medical Condition"] = df.groupby("CPT_HCPCS_Code")["Medical Condition"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))

df["Medical Condition"] = (df.groupby("CPT_HCPCS_Code")["Medical Condition"].transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else "Unknown")))

df["Medical Condition"] = (df.groupby("Procedure_Description")["Medical Condition"].transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else "Unknown")))

df["Medical Condition"].isnull().sum()
```

```python
# CPT_HCPCS_Code – Analyze the missing values based on Medical Condition.

df.loc[df["CPT_HCPCS_Code"].isnull(),"Medical Condition"].value_counts()

df["CPT_HCPCS_Code"].isnull().sum()

```

```python
# CPT_HCPCS_Code – Replace Null Values Using the Mode Based on Medical Condition.

df["CPT_HCPCS_Code"] = df.groupby("Medical Condition")["CPT_HCPCS_Code"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))

df["CPT_HCPCS_Code"].isnull().sum()

df["CPT_HCPCS_Code"] = df["CPT_HCPCS_Code"].astype(int)
```

```python
# Procedure_Description – Analyze the missing values based on CPT_HCPCS_Code.

df.loc[df["Procedure_Description"].isnull(),"CPT_HCPCS_Code"].value_counts()
```

```python
# Procedure_Description – Replace Null Values Using the Mode Based on CPT_HCPCS_Code.

df["Procedure_Description"] = df.groupby("CPT_HCPCS_Code")["Procedure_Description"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))

df["Procedure_Description"].isnull().sum()
```

```python
# Provider_ID – Analyze the missing values based on Provider_Name.

df.loc[df["Provider_ID"].isnull(),("Provider_Name")].value_counts()
```

```python
# Provider_ID – Replace Null Values Using the Mode Based on Provider_Name.

df["Provider_ID"] = df.groupby("Provider_Name")["Provider_ID"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))

df["Provider_ID"].isnull().sum()
```

```python
# Provider_Name – Analyze the missing values based on Provider_ID.

df.loc[df["Provider_Name"].isnull(),"Provider_ID"].value_counts()
```

```python
# Provider_Name – Replace Null Values Using the Mode Based on Provider_ID.

df["Provider_Name"] = df.groupby("Provider_ID")["Provider_Name"].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else "Unknown"))

df["Provider_Name"].isnull().sum()
```

```python
# Claim_Month – Replace null values using the Discharge Date column by extracting the Year and Month.

df["Claim_Month"] = df["Claim_Month"].fillna(pd.to_datetime(df["Discharge Date"],dayfirst=True).dt.strftime("%Y-%m"))

df["Claim_Month"].isnull().sum()
```

```python
# Claim_Year – Analysing the null values using the Discharge Date column by extracting the Year.

df.iloc[24,18]

df["Claim_Year"].isnull().sum()
```

```python
# Claim_Year – Replace null values using the Discharge Date column by extracting the Year.

#df["Claim_Year"] = df["Claim_Year"].fillna(pd.to_datetime(df["Discharge Date"], dayfirst=True).dt.strftime("%Y"))

df["Claim_Year"].isnull().sum()
```

```python
# Allowed_Amount – Replace null values using the Billed_Amount and Contractual_Adjustment columns.

df["Allowed_Amount"] = df["Allowed_Amount"].fillna(df["Billed_Amount"] - df["Contractual_Adjustment"])

df["Allowed_Amount"].isnull().sum()
```

```python
# Outstanding_Amount – Replace null values using the Allowed_Amount and Paid_Amount columns.


df["Outstanding_Amount"] = df["Outstanding_Amount"].fillna(df["Allowed_Amount"] - df["Paid_Amount"]).round(2)

df["Allowed_Amount"].isnull().sum()
```

```python
# Create a mapping between Denial_Reason and the correct Denial_Category

denial_category_mapping = {
    "Precertification/authorization absent": "Authorization",
    "Charge exceeds fee schedule/contracted rate": "Contractual",
    "Non-covered service - not medically necessary": "Eligibility",
    "Deductible amount": "Eligibility",
    "Benefit included in another service already paid": "Technical",
    "Timely filing limit expired": "Technical",
    "Claim lacks information needed for adjudication": "Technical",
    "Duplicate claim/service": "Technical"
}

# Update the Denial_Category column based on Denial_Reason
df["Denial_Category"] = df["Denial_Reason"].map(denial_category_mapping)
```

```python
ar_counts = df["AR_Aging_Bucket"].value_counts()

aging_order = [
    "0-30",
    "31-60",
    "61-90",
    "91-120",
    "121-180",
    "181-365",
    "365+"
]

ar_counts = ar_counts.reindex(aging_order).dropna()

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    y=ar_counts.index,
    x=ar_counts.values
)

for i, value in enumerate(ar_counts.values):
    ax.text(value, i, f" {int(value)}", va="center")

plt.title("Distribution of Claims by AR Aging Bucket")
plt.xlabel("Number of Claims")
plt.ylabel("AR Aging Bucket")

plt.show()
```

```python
denial_counts = df["Denial_Category"].value_counts()

def show_count_and_percent(pct):
    count = int(round(pct / 100 * denial_counts.sum()))
    return f"{count}\n({pct:.1f}%)"

plt.figure(figsize=(8, 8))

plt.pie(
    denial_counts,
    labels=denial_counts.index,
    autopct=show_count_and_percent,
    startangle=90
)

plt.title("Distribution of Denial Categories")
plt.show()
```
